# 1. Imports and paths

```python
from pathlib import Path
import pandas as pd

REPO_ROOT = Path.cwd().parent

CONFIG = "HC-EWI"

data_file = REPO_ROOT / "data" / "sample" / CONFIG / "Data_sample.txt"
model_file = REPO_ROOT / "data" / "sample" / CONFIG / "model_results_sample.csv"

print(f"Repository: {REPO_ROOT}")
print(f"Configuration: {CONFIG}")
print(f"Data file exists: {data_file.exists()}")
print(f"Model-results file exists: {model_file.exists()}")
```

# 2. Raw data sample

## 2.1 Inspection

```python
data = pd.read_csv(
    data_file,
    sep=",",
    encoding="utf-8-sig",
)
data.head()
```

```python
print("Columns:")
for column in data.columns:
    print(f"  - {column}")
```

```python
print(f"Shape: {data.shape}")
```

## 2.2 Checking for missing data

```python
missing = data.isna().sum()
missing[missing > 0]
```

# 3. Model results data

```python
model_results = pd.read_csv(
    model_file,
    sep=",",
    encoding="utf-8-sig",
)

print(f"Shape: {model_results.shape}")
model_results.head()
```

```python
print("Raw data:")
print(f"  Rows:    {len(data):,}")
print(f"  Columns: {len(data.columns)}")

print("\nModel results:")
print(f"  Rows:    {len(model_results):,}")
print(f"  Columns: {len(model_results.columns)}")
```
